# Netflix Titles — Exploratory Data AnalysisAnalyzing the Netflix catalog (titles added through 2021) to understand content mix, growth over time, geography, ratings, and genres.

## 1. Setup & Import

```python
import numpy as npimport pandas as pdimport matplotlib.pyplot as pltimport seaborn as snssns.set_style('whitegrid')
```

## 2. Load Data

```python
df = pd.read_csv("netflix_titles_2021.csv")df.head()
```

**Missing data notes:**- `director` is missing for a large share of rows (~30%)- `cast` and `country` each have a modest number of gaps- `date_added`, `rating`, and `duration` are missing only a handful of rows

## 4. Data Cleaning & Feature Engineering

```python
df1 = df.copy()# Drop rows without a date_added since we need it for the time-based analysisdf1.dropna(subset=['date_added'], inplace=True)# Fill categorical gaps with an explicit 'unknown' label instead of dropping rowsdf1['director'] = df1['director'].fillna('unknown')df1['cast'] = df1['cast'].fillna('unknown')df1['country'] = df1['country'].fillna('unknown')# Parse date_added and extract the year the title was added to Netflixdf1['date_added'] = df1['date_added'].astype(str).str.strip()df1['date_added'] = pd.to_datetime(df1['date_added'], format='%B %d, %Y', errors='coerce')df1.dropna(subset=['date_added'], inplace=True)df1['year_added'] = df1['date_added'].dt.yeardf1.shape
```

## 5. Exploratory Analysis### Q1. Movies vs. TV Shows — what's the split?

```python
plt.figure(figsize=(6,4))sns.countplot(x='type', data=df1)plt.title('Content Type: Movies vs. TV Shows')plt.xlabel('Type')plt.ylabel('Number of Titles')plt.tight_layout()plt.show()
```

### Q2. Which countries produce the most Netflix content?

```python
plt.figure(figsize=(8,5))df1['country'].value_counts().head(10).plot(kind='bar')plt.title('Top 10 Countries by Number of Titles')plt.xlabel('Country')plt.ylabel('Number of Titles')plt.tight_layout()plt.show()
```

### Q3. How has the number of titles added grown over time?

```python
plt.figure(figsize=(8,5))df1['year_added'].value_counts().sort_index().plot(marker='o')plt.title('Titles Added to Netflix by Year')plt.xlabel('Year Added')plt.ylabel('Number of Titles')plt.tight_layout()plt.show()
```

### Q4. What are the most common content ratings?

```python
plt.figure(figsize=(9,5))df1['rating'].value_counts().plot(kind='bar')plt.title('Number of Titles by Content Rating')plt.xlabel('Rating')plt.ylabel('Number of Titles')plt.tight_layout()plt.show()
```

### Q5. What are the most common genres/categories?

```python
df1['listed_in'].value_counts().head(10)
```

### Q6. Do Movies and TV Shows follow the same growth pattern?

```python
movies = df1[df1['type'] == 'Movie']tv = df1[df1['type'] == 'TV Show']plt.figure(figsize=(8,5))movies['year_added'].value_counts().sort_index().plot(marker='o')plt.title('Movies Added to Netflix by Year')plt.xlabel('Year Added')plt.ylabel('Number of Movies')plt.tight_layout()plt.show()
```

```python
plt.figure(figsize=(8,5))tv['year_added'].value_counts().sort_index().plot(marker='o', color='orange')plt.title('TV Shows Added to Netflix by Year')plt.xlabel('Year Added')plt.ylabel('Number of TV Shows')plt.tight_layout()plt.show()
```

## 6. Key Insights- Movies make up roughly 70% of the catalog vs. ~30% for TV shows- The United States and India are the largest content sources by far- Catalog growth was flat before 2015, then rose sharply from 2016–2019, and dipped in 2020–2021- TV-MA and TV-14 are the two most common ratings, indicating a mostly mature-audience catalog- International dramas, documentaries, and stand-up comedy are the leading genre combinations- Movies and TV shows follow a similar growth curve, but movie additions consistently outpace TV show additions in volume
